```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Part-1:Data Generation.

```python
# Generate 3000 wait times for regular days using uniform distribution (2 to 12 minutes)
regular_wait_times = np.random.uniform(2, 12, 3000)

# Generate 3000 wait times for busy days using normal distribution (mean=8, std=2)
busy_wait_times = np.random.normal(8, 2, 3000)

# Replace any negative wait times with 0
busy_wait_times = np.where(busy_wait_times < 0, 0, busy_wait_times)

print("Regular day sample:", regular_wait_times)
print("Busy day sample:", busy_wait_times)
```

# Part-2:Data Visualization.

```python
# Create a figure with 2 subplots side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 5))  # figure object, subplot array

# Histogram for Regular Days
axes[0].hist(regular_wait_times, bins=30, alpha=0.75, color='blue', edgecolor='black')
axes[0].set_title('Regular Days - Wait Time (Uniform Distribution)')
axes[0].set_xlabel('Wait Time (minutes)')
axes[0].set_ylabel('Number of Customers')
axes[0].grid(True)

# Histogram for Busy Days
axes[1].hist(busy_wait_times, bins=30, alpha=0.75, color='red', edgecolor='black')
axes[1].set_title('Busy Days - Wait Time (Normal Distribution)')
axes[1].set_xlabel('Wait Time (minutes)')
axes[1].set_ylabel('Number of Customers')
axes[1].grid(True)


plt.tight_layout()  # adjusts spacing so titles/labels don't overlap
plt.show()
```

# OR: Overlaid with transparency

```python
# Plot both distributions on the same axes using alpha for transparency
plt.hist(regular_wait_times, bins=30, alpha=0.5, color='blue', edgecolor='black', label='Regular Days')
plt.hist(busy_wait_times, bins=30, alpha=0.5, color='red', edgecolor='black', label='Busy Days')

plt.xlabel('Wait Time (minutes)')
plt.ylabel('Number of Customers')
plt.title('Overlaid Histogram of Wait Times')
plt.legend()
plt.grid(True)
plt.show()
```

# Part-3:Statistical Analysis

```python
# Statistics for Regular Days
regular_mean = np.mean(regular_wait_times)
regular_median = np.median(regular_wait_times)
regular_std = np.std(regular_wait_times)
regular_max = np.max(regular_wait_times)

# Statistics for Busy Days
busy_mean = np.mean(busy_wait_times)
busy_median = np.median(busy_wait_times)
busy_std = np.std(busy_wait_times)
busy_max = np.max(busy_wait_times)

print("Regular Days (Uniform Distribution)")
print("Mean:", regular_mean)
print("Median:", regular_median)
print("Standard Deviation:", regular_std)
print("Maximum Wait Time:", regular_max)

print("\nBusy Days (Normal Distribution)")
print("Mean:", busy_mean)
print("Median:", busy_median)
print("Standard Deviation:", busy_std)
print("Maximum Wait Time:", busy_max)
```
