```python
import math
import random
import numpy as np
import pandas as pd
```

### Part 1 - Setup & Personalization

```python
# Student ID: 47
R = 47

# Range for x and y coordinates: [0, R/10]
range_max = R / 10
print(f"Random numbers will be generated in the range [0, {range_max}]")

# Circle radius: R/18
circle_radius = R / 18
print(f"The circle radius is {circle_radius}")

# Number of random points to generate
full_name = "Monika Akter"
number_of_letters = sum(c.isalpha() for c in full_name)
initial_total_points = 200 * number_of_letters
print(f"Number of letters in your full name: {number_of_letters}")
print(f"Initial total points to generate: {initial_total_points}")
```

###Part 2 - Simulation Logic

```python
def is_inside_circle(x,y,circle_radius):

    distance_squared=x**2+y**2

    if distance_squared <=circle_radius**2:
        return True
    else:
        return False
def run_simulation(total_points):

    points_inside=0
    points_outside=0

    half_points=total_points//2

    for i in range(half_points):


        x = random.uniform(0,range_max)
        y = random.uniform(0,range_max)
        if is_inside_circle(x,y,circle_radius):
            points_inside +=1
        else:
            points_outside +=1
    remaining_points=total_points-half_points

    for i in range(remaining_points):

        x = np.random.rand()*range_max
        y = np.random.rand()*range_max

        if is_inside_circle(x,y,circle_radius):
            points_inside +=1
        else:
            points_outside +=1

    estimated_pi=4*points_inside/total_points
    error=abs(estimated_pi-math.pi)

    return points_inside,points_outside,estimated_pi,error
```

### Part 3 - Output & Analysis

run the simulation 5 times with increasing numbers of points and compare the errors.

```python
point_values = [
    initial_total_points,
    initial_total_points*2,
    initial_total_points*3,
    initial_total_points*4,
    initial_total_points*5
]



print("SIMULATION RESULTS")
print(
    f"{'Range Used':<15}"
    f"{'Total Points':<15}"
    f"{'Points Inside':<17}"
    f"{'Estimated PI':<17}"
    f"{'Error from math.pi':<20}"
)

for total_points in point_values:

    inside,outside,estimated_pi,error=run_simulation(total_points)
    print(
        f"[0,{range_max}]".ljust(15),
        f"{total_points:<15}",
        f"{inside:<17}",
        f"{estimated_pi:<17.6f}",
        f"{error:<20.6f}"
    )
```
