# Task-1,2:Create personal matrix A then Analyze matrix A.

```python
import numpy as np

# Student ID: 1047
d1 = 7
d2 = 4

# Matrix A
A = np.array([[d1+2, d2+1],[2*d1, d2+2]])

print("Matrix A:\n", A)

# Shape
print("\nShape of A:", A.shape)

# Determinant
det_A = np.linalg.det(A)
print("Determinant of A:",det_A)

# Rank
rank_A = np.linalg.matrix_rank(A)
print("Rank of A:", rank_A)

# Eigenvalues
eigen_A = np.linalg.eigvals(A)
print("Eigenvalues of A:", eigen_A)

# Inverse
if det_A != 0:
    inverse_A = np.linalg.inv(A)
    print("Inverse of A:",inverse_A)
else:
    print("A has no inverse")
```

# Task-3,4:Create New matrix B then Re-analyze matrix B.

```python
# Changing exactly one value
# A[0,0] = 9 -> 10

B = A.copy()
B[0, 0] = B[0, 0] + 1

print("\n\nChanged element: A[0,0] = 9 to B[0,0] = 10")

print("\nMatrix B:", B)

# Shape
print("\nShape of B:", B.shape)

# Determinant
det_B = np.linalg.det(B)
print("Determinant of B:", det_B)

# Rank
rank_B = np.linalg.matrix_rank(B)
print("Rank of B:", rank_B)

# Eigenvalues
eigen_B = np.linalg.eigvals(B)
print("Eigenvalues of B:",eigen_B)

# Inverse
if det_B != 0:
    inverse_B = np.linalg.inv(B)
    print("Inverse of B:",inverse_B)
else:
    print("B has no inverse")
```

1.How did the determinant change and why?

  Ans: The determinant changed from -16.0 for Matrix A to -10.0 for Matrix B. The determinant increased. This change happens because the determinant depends on all the elements in the matrix. When you change a single element, it directly affects the calculation of the determinant. For a 2x2 matrix [[a, b], [c, d]], the determinant is ad-bc. When A[0,0], which is a, is increased by 1, the ad term increases. This leads to a change in the overall determinant.

2.Did the rank change? Explain.

  Ans: No,there was no change in rank. The rank of both Matrix A and Matrix B is 2. If the determinant of a 2x2 matrix is not zero, the rank is 2. Both matrices are full rank (rank 2) since the determinants of A (-16.0) and B (-10.0) are non-zero.
  
3.How did the eigenvalues respond to the value change?
    
  Ans: Significant changes occurred in the eigenvalues. The eigenvalues of Matrix A were [16.0, -1.0]. Their approximate values for Matrix B were [16.6023, -0.6023]. The value of both eigenvalues changed. The characteristic polynomial, which is formed from the matrix elements, has eigenvalues as its roots. This polynomial's roots, or eigenvalues, will change if even one of its elements is altered. In this specific case, one eigenvalue increased and the other became less negative (also an increase).

4.Is B easier or harder to invert than A? Why?

 Ans:Based on this modification, neither Matrix B nor Matrix A are inherently "easier" or "harder" to invert. Both matrices are invertible since their determinants are non-zero (-16.0 and -10.0, respectively). As long as the determinant is non-zero, the computational complexity of computing the inverse often depends on the size of the matrix rather than the particular values. To be invertible, the determinant must simply not be zero. Since both determinants are non-zero, both inverses exist and can be calculated using conventional techniques (such as the adjugate matrix), which, under the assumption of similar numerical stability, do not become "easier" or "harder" by a single element change.
